````python
import pandas as pd

df = pd.read_csv(
    "/content/WELFake_Dataset.csv",
    encoding="latin-1",
    header=None
)

print("Shape:", df.shape)
print(df.iloc[:5, :5])
````

````python
#CLEAN THE DATASET

# Keep only the useful columns
df = df.iloc[:, :4].copy()

# Set the correct column names
df.columns = ["id", "title", "text", "label"]

# Remove the header row that was read as data
df = df[df["id"] != "id"].copy()

# Reset the index
df.reset_index(drop=True, inplace=True)

print("Dataset shape:", df.shape)
print(df.head())
````

````python
print(df["label"].value_counts())
````

````python
print(df["label"].unique())
````

````python
print(df.isnull().sum())
````

````python
# Fill missing titles and article text
df["title"] = df["title"].fillna("")
df["text"] = df["text"].fillna("")

# Make sure labels are numeric
df["label"] = pd.to_numeric(df["label"], errors="coerce")

# Remove rows where label is missing
df = df.dropna(subset=["label"])

# Convert labels to integers
df["label"] = df["label"].astype(int)

print("Dataset shape:", df.shape)
print(df.isnull().sum())
````

````python
df["content"] = df["title"] + " " + df["text"]

print(df[["content", "label"]].head())
````

````python
df = df[df["content"].str.strip() != ""].copy()

print("Remaining articles:", len(df))
````

````python
print(df["label"].value_counts())
````

````python
print(df["label"].value_counts(dropna=False))
````

````python
print(df["label"].unique())
````

````python
# Keep only valid binary labels: 0 and 1
df = df[df["label"].isin([0, 1])].copy()

# Make sure labels are integers
df["label"] = df["label"].astype(int)

print("Clean dataset shape:", df.shape)
print("Label distribution:")
print(df["label"].value_counts())
````

````python
from sklearn.model_selection import train_test_split

X = df["content"]
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
````

````python
from tensorflow.keras.preprocessing.text import Tokenizer

MAX_WORDS = 20000

tokenizer = Tokenizer(
    num_words=MAX_WORDS,
    oov_token="<OOV>"
)

tokenizer.fit_on_texts(X_train)

print("Vocabulary size:", len(tokenizer.word_index))
````

````python
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq = tokenizer.texts_to_sequences(X_test)

print(X_train_seq[0][:20])
````

````python
from tensorflow.keras.preprocessing.sequence import pad_sequences

MAX_LEN = 300

X_train_pad = pad_sequences(
    X_train_seq,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

X_test_pad = pad_sequences(
    X_test_seq,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

print("Training shape:", X_train_pad.shape)
print("Testing shape:", X_test_pad.shape)
````

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense, Dropout

model = Sequential([
    Input(shape=(MAX_LEN,)),

    Embedding(
        input_dim=MAX_WORDS,
        output_dim=128
    ),

    SimpleRNN(64),

    Dropout(0.3),

    Dense(1, activation="sigmoid")
])

model.summary()
````

````python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully.")
````

````python
history = model.fit(
    X_train_pad,
    y_train,
    validation_split=0.20,
    epochs=3,
    batch_size=64,
    verbose=1
)
````

````python
test_loss, test_accuracy = model.evaluate(
    X_test_pad,
    y_test,
    verbose=1
)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
````

````python
from sklearn.metrics import classification_report

y_probability = model.predict(X_test_pad)

y_pred = (y_probability >= 0.5).astype(int).flatten()

print(classification_report(
    y_test,
    y_pred,
    digits=4
))
````

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, LSTM, Dense, Dropout

lstm_model = Sequential([
    Input(shape=(MAX_LEN,)),

    Embedding(
        input_dim=MAX_WORDS,
        output_dim=128
    ),

    LSTM(64),

    Dropout(0.3),

    Dense(1, activation="sigmoid")
])

lstm_model.summary()
````

````python
lstm_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("LSTM model compiled successfully.")
````

````python
lstm_history = lstm_model.fit(
    X_train_pad,
    y_train,
    validation_split=0.20,
    epochs=3,
    batch_size=64,
    verbose=1
)
````

````python
test_loss, test_accuracy = lstm_model.evaluate(
    X_test_pad,
    y_test,
    verbose=1
)

print("LSTM Test Loss:", test_loss)
print("LSTM Test Accuracy:", test_accuracy)
````

````python
from sklearn.metrics import classification_report

lstm_probability = lstm_model.predict(X_test_pad)

lstm_pred = (lstm_probability >= 0.5).astype(int).flatten()

print(classification_report(
    y_test,
    lstm_pred,
    digits=4
))
````

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

better_lstm = Sequential([
    Input(shape=(MAX_LEN,)),

    Embedding(
        input_dim=MAX_WORDS,
        output_dim=128
    ),

    LSTM(64),

    Dropout(0.4),

    Dense(1, activation="sigmoid")
])

better_lstm.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=1,
    restore_best_weights=True
)

better_history = better_lstm.fit(
    X_train_pad,
    y_train,
    validation_split=0.20,
    epochs=5,
    batch_size=64,
    callbacks=[early_stopping],
    verbose=1
)
````

````python
better_test_loss, better_test_accuracy = better_lstm.evaluate(
    X_test_pad,
    y_test,
    verbose=1
)

print("Better LSTM Test Accuracy:", better_test_accuracy)
````

````python
better_probability = better_lstm.predict(X_test_pad)

better_pred = (
    better_probability >= 0.5
).astype(int).flatten()

print(classification_report(
    y_test,
    better_pred,
    digits=4
))
````

````python
def predict_news(article):

    sequence = tokenizer.texts_to_sequences([article])

    padded = pad_sequences(
        sequence,
        maxlen=MAX_LEN,
        padding="post",
        truncating="post"
    )

    probability = float(
        better_lstm.predict(padded, verbose=0)[0][0]
    )

    prediction = 1 if probability >= 0.5 else 0

    return prediction, probability
````
